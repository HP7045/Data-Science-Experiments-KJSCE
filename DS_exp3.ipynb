Cell 1 — Import libraries and select Excel file

```python
import numpy as np
import pandas as pd
import tkinter as tk
from tkinter import filedialog

root = tk.Tk()
root.withdraw()

file_path = filedialog.askopenfilename(
    title="Select Weather Dataset",
    filetypes=[
        ("Excel files", "*.xlsx *.xls"),
        ("All files", "*.*")
    ]
)

root.destroy()

if not file_path:
    raise FileNotFoundError("No dataset was selected.")

df = pd.read_excel(file_path)

numeric_columns = [
    "temperature_c",
    "relative_humidity",
    "wind_speed",
    "precipitation_amount",
    "cloud_cover",
    "lifted_index"
]

for col in numeric_columns:
    df.fillna({col: df[col].median()}, inplace=True)

print("Dataset loaded successfully.")
print("File:", file_path)
print("Dataset shape:", df.shape)

print("\nColumns analyzed:")
print(numeric_columns)

print(
    "\nObservation: The weather dataset was successfully "
    "loaded and numerical weather attributes were selected "
    "for statistical analysis."
)
```

Cell 2 — Manual statistical functions

```python
def manual_mean(data):
    data = list(data)
    total = 0

    for value in data:
        total += value

    return total / len(data)


def manual_median(data):
    data = sorted(list(data))
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]

    return (data[n // 2 - 1] + data[n // 2]) / 2


def manual_mode(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    maximum_frequency = 0
    modes = []

    for value in frequency:
        if frequency[value] > maximum_frequency:
            maximum_frequency = frequency[value]
            modes = [value]

        elif frequency[value] == maximum_frequency:
            modes.append(value)

    return modes, maximum_frequency


def manual_variance(data):
    data = list(data)
    mean = manual_mean(data)
    total = 0

    for value in data:
        total += (value - mean) ** 2

    return total / (len(data) - 1)


def manual_standard_deviation(data):
    return manual_variance(data) ** 0.5


def manual_quartile(data, position):
    data = sorted(list(data))
    n = len(data)

    if position == 1:
        lower = data[:n // 2]
        return manual_median(lower)

    if position == 3:
        upper = data[(n + 1) // 2:]
        return manual_median(upper)


def manual_iqr(data):
    q1 = manual_quartile(data, 1)
    q3 = manual_quartile(data, 3)

    return q3 - q1


def manual_frequency(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

    return frequency


print("Manual statistical functions created successfully.")

print(
    "Observation: Mean, mode, median, variance, standard deviation, "
    "quartiles, IQR and frequency calculations were implemented manually."
)
```

Cell 3 — Statistical analysis of each numerical attribute

```python
for column in numeric_columns:

    data = df[column].dropna().tolist()

    mean_value = manual_mean(data)
    median_value = manual_median(data)
    mode_values, mode_frequency = manual_mode(data)
    variance_value = manual_variance(data)
    standard_deviation = manual_standard_deviation(data)

    q1 = manual_quartile(data, 1)
    q3 = manual_quartile(data, 3)
    iqr_value = manual_iqr(data)

    print("\n" + "=" * 75)
    print("STATISTICS FOR:", column)
    print("=" * 75)

    print("Frequency:", len(data))
    print("Mean:", mean_value)
    print("Mode:", mode_values)
    print("Mode Frequency:", mode_frequency)
    print("Median:", median_value)
    print("Variance:", variance_value)
    print("Standard Deviation:", standard_deviation)
    print("Q1:", q1)
    print("Q3:", q3)
    print("Interquartile Range:", iqr_value)

    print(
        "\nObservation: The statistical measures for",
        column,
        "describe its central tendency and dispersion."
    )
```

Cell 4 — Frequency analysis

```python
print("\n" + "=" * 75)
print("WEATHER FREQUENCY")
print("=" * 75)

weather_frequency = manual_frequency(
    df["weather"].dropna().tolist()
)

for value in weather_frequency:
    print(value, ":", weather_frequency[value])

print(
    "\nObservation: The frequency distribution shows how often "
    "each weather condition occurs in the dataset."
)


print("\n" + "=" * 75)
print("PRECIPITATION TYPE FREQUENCY")
print("=" * 75)

precipitation_frequency = manual_frequency(
    df["precipitation_type"].dropna().tolist()
)

for value in precipitation_frequency:
    print(value, ":", precipitation_frequency[value])

print(
    "\nObservation: The precipitation frequency distribution "
    "shows the occurrence of different precipitation types."
)
```

Cell 5 — Grouped data analysis by city

```python
cities = df["city"].dropna().unique()

for city in cities:

    city_data = df[df["city"] == city]

    print("\n" + "-" * 75)
    print("CITY:", city)
    print("-" * 75)

    for column in numeric_columns:

        data = city_data[column].dropna().tolist()

        if len(data) < 2:
            continue

        mean_value = manual_mean(data)
        median_value = manual_median(data)
        mode_values, mode_frequency = manual_mode(data)
        variance_value = manual_variance(data)
        standard_deviation = manual_standard_deviation(data)

        q1 = manual_quartile(data, 1)
        q3 = manual_quartile(data, 3)
        iqr_value = manual_iqr(data)

        print("\n", column)
        print("Frequency:", len(data))
        print("Mean:", mean_value)
        print("Mode:", mode_values)
        print("Mode Frequency:", mode_frequency)
        print("Median:", median_value)
        print("Variance:", variance_value)
        print("Standard Deviation:", standard_deviation)
        print("Q1:", q1)
        print("Q3:", q3)
        print("Interquartile Range:", iqr_value)

    print(
        "\nObservation: The numerical weather attributes were "
        "analyzed separately for",
        city,
        "to compare their statistical characteristics."
    )
```

Cell 6 — Grouped frequency analysis

```python
print("\n" + "=" * 75)
print("GROUPED WEATHER FREQUENCY BY CITY")
print("=" * 75)

grouped_weather = df.groupby(
    ["city", "weather"]
).size()

print(grouped_weather)

print(
    "\nObservation: Weather conditions were grouped by city "
    "and their frequencies were calculated to determine the "
    "distribution of weather conditions across different cities."
)
```

Cell 7 — Grouped temperature analysis

```python
print("\n" + "=" * 75)
print("GROUPED TEMPERATURE ANALYSIS")
print("=" * 75)

grouped_temperature = {}

for city in cities:

    data = df[
        df["city"] == city
    ]["temperature_c"].dropna().tolist()

    if len(data) > 1:

        grouped_temperature[city] = {
            "Frequency": len(data),
            "Mean": manual_mean(data),
            "Mode": manual_mode(data)[0],
            "Median": manual_median(data),
            "Variance": manual_variance(data),
            "Standard Deviation": manual_standard_deviation(data),
            "Q1": manual_quartile(data, 1),
            "Q3": manual_quartile(data, 3),
            "IQR": manual_iqr(data)
        }

grouped_temperature_df = pd.DataFrame.from_dict(
    grouped_temperature,
    orient="index"
)

display(grouped_temperature_df)

print(
    "\nObservation: Temperature statistics were calculated "
    "separately for each city using the manual statistical procedures."
)
```

Cell 8 — Highest and lowest mean temperature

```python
highest_mean_city = None
lowest_mean_city = None

highest_mean = None
lowest_mean = None

for city in grouped_temperature:

    mean_value = grouped_temperature[city]["Mean"]

    if highest_mean is None or mean_value > highest_mean:
        highest_mean = mean_value
        highest_mean_city = city

    if lowest_mean is None or mean_value < lowest_mean:
        lowest_mean = mean_value
        lowest_mean_city = city


print("\n" + "=" * 75)
print("TEMPERATURE COMPARISON")
print("=" * 75)

print(
    "\nCity with highest mean temperature:",
    highest_mean_city
)

print(
    "Mean temperature:",
    highest_mean
)

print(
    "\nCity with lowest mean temperature:",
    lowest_mean_city
)

print(
    "Mean temperature:",
    lowest_mean
)

print(
    "\nObservation: The cities with the highest and lowest "
    "average temperatures were identified by comparing their "
    "manually calculated mean values."
)
```

Cell 9 — Outlier analysis

```python
print("\n" + "=" * 75)
print("OUTLIER ANALYSIS FOR EACH NUMERICAL ATTRIBUTE")
print("=" * 75)

for column in numeric_columns:

    data = df[column].dropna().tolist()

    q1 = manual_quartile(data, 1)
    q3 = manual_quartile(data, 3)

    iqr_value = q3 - q1

    lower_bound = q1 - (1.5 * iqr_value)
    upper_bound = q3 + (1.5 * iqr_value)

    outliers = []

    for value in data:

        if value < lower_bound or value > upper_bound:
            outliers.append(value)

    print("\n" + "-" * 75)
    print("ATTRIBUTE:", column)
    print("-" * 75)

    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr_value)

    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)

    print("Number of Outliers:", len(outliers))

    print("\nOutlier Values:")

    if len(outliers) > 0:
        print(outliers)
    else:
        print("No outliers detected.")

    print(
        "\nObservation: Values below the lower bound or above "
        "the upper bound were classified as outliers using "
        "the 1.5 × IQR rule."
    )
```

Cell 10 — Outlier records with city information

```python
print("\n" + "=" * 75)
print("OUTLIER RECORDS WITH CITY INFORMATION")
print("=" * 75)

for column in numeric_columns:

    data = df[column].dropna().tolist()

    q1 = manual_quartile(data, 1)
    q3 = manual_quartile(data, 3)

    iqr_value = q3 - q1

    lower_bound = q1 - (1.5 * iqr_value)
    upper_bound = q3 + (1.5 * iqr_value)

    outlier_records = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print("\n" + "-" * 75)
    print("OUTLIERS FOR:", column)
    print("-" * 75)

    if len(outlier_records) > 0:

        display(
            outlier_records[
                ["city", column]
            ]
        )

        print(
            "Number of outlier records:",
            len(outlier_records)
        )

    else:

        print("No outlier records detected.")

    print(
        "\nObservation: The corresponding cities and values "
        "associated with detected outliers were displayed "
        "for further analysis."
    )
```
